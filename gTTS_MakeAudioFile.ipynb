```python
# Install required packages
!pip install -q python-docx gTTS pydub

from docx import Document
from gtts import gTTS
from pydub import AudioSegment
from google.colab import files
import os
import re

# ============================================================
# 1. File name
# ============================================================

input_file = "Portugali_1_English.docx"

# ============================================================
# 2. Read the Word document
# ============================================================

doc = Document(input_file)

paragraphs = []

for paragraph in doc.paragraphs:
    text = paragraph.text.strip()
    if text:
        paragraphs.append(text)

text = "\n\n".join(paragraphs)

print("Document loaded.")
print("Number of characters:", len(text))
print("Number of paragraphs:", len(paragraphs))

# ============================================================
# 3. Split into manageable chunks
# ============================================================

max_chars = 4000

chunks = []

for i in range(0, len(text), max_chars):
    chunk = text[i:i + max_chars].strip()
    if chunk:
        chunks.append(chunk)

print("Number of audio parts:", len(chunks))

# ============================================================
# 4. Convert each chunk to speech
# ============================================================

audio_files = []

for i, chunk in enumerate(chunks):

    filename = f"Portugali_English_part_{i+1}.mp3"

    print(f"Creating part {i+1}/{len(chunks)}...")

    tts = gTTS(
        text=chunk,
        lang="en",
        tld="co.uk", #set variant us:tld="com"  , australia: tld="com.au", indian tld="co.in"
        slow=False #this set speed to normal
    )

    tts.save(filename)
    audio_files.append(filename)

print("All parts created.")

# ============================================================
# 5. Merge all MP3 files into ONE audio file
# ============================================================

combined = AudioSegment.empty()

for filename in audio_files:
    audio = AudioSegment.from_mp3(filename)
    combined += audio

output_file = "Portugali_1_English_FULL.mp3"

combined.export(
    output_file,
    format="mp3"
)

print("Finished!")
print("Output:", output_file)

# ============================================================
# 6. Download the final audio
# ============================================================

files.download(output_file)
```

```python
#make sound faster


from pydub import AudioSegment

audio = AudioSegment.from_mp3("Portugali_1_English_FULL.mp3")

# 1.25 = 25% faster
faster = audio.speedup(playback_speed=1.25)

faster.export(
    "Portugali_1_English_1.25x.mp3",
    format="mp3"
)
```
